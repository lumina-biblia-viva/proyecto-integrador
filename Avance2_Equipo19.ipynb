### **2.1 Carga del dataset y preprocesamiento mínimo**

En esta sección se realiza la **carga del corpus estructurado** y un **preprocesamiento mínimo y consciente**, alineado con el enfoque de *LUMINA*: preservar la fidelidad textual y la trazabilidad del contenido bíblico.

A diferencia de pipelines típicos de NLP, aquí **no se aplica lematización, stemming ni normalización agresiva**, ya que dichas transformaciones podrían alterar matices semánticos importantes para el objetivo del sistema RAG. En su lugar, se aplican únicamente operaciones seguras y reversibles, como:

- Normalización de espacios en blanco (colapsar espacios múltiples).
- Conversión controlada de valores nulos a cadenas vacías.
- Verificaciones básicas de integridad en columnas textuales.

Además, se definen las rutas actualizadas para acceder al dataset en Google Drive y para guardar los artefactos generados en el avance anterior (EDA).

### **2.2 Features estructurales (metadatos canónicos) y numéricas auxiliares**

En esta sección se construyen **características estructurales y numéricas auxiliares** que complementan la representación semántica del texto bíblico. Estas características no buscan sustituir los embeddings, sino **aportar contexto canónico y señales cuantitativas que permitan**:

- Mitigar sesgos estructurales identificados en el EDA.
- Facilitar reglas posteriores de segmentación (chunking).
- Mejorar la interpretabilidad y trazabilidad del sistema.

Se incluyen:

- **Metadatos canónicos derivados**, como testamento y tipo de libro.
- **Variables numéricas auxiliares**, como longitud del versículo y diferencias entre traducciones.
- **Discretización (binning)** de longitudes, aplicada únicamente a variables numéricas, nunca al texto crudo.

Todas las decisiones están fundamentadas en los hallazgos del Análisis Exploratorio de Datos.

```python
# ============================================================
# Parte 2.2 — Features estructurales y numéricas auxiliares
# ============================================================

# -----------------------------
# 1) Features estructurales: Testamento
# -----------------------------
# Inferimos el testamento a partir del orden canónico detectado en el dataset.
# (39 libros AT, 27 NT)

books_in_order = (
    df[["book", "book_es"]]
    .drop_duplicates()
    .reset_index(drop=True)
)

if len(books_in_order) != 66:
    print("Advertencia: se esperaban 66 libros, se detectaron:", len(books_in_order))

books_in_order["testament"] = (
    ["AT"] * min(39, len(books_in_order)) +
    ["NT"] * max(0, len(books_in_order) - 39)
)

book_to_testament = dict(zip(books_in_order["book"], books_in_order["testament"]))
df["testament"] = df["book"].map(book_to_testament)

print("Distribución por testamento:")
print(df["testament"].value_counts())

# -----------------------------
# 2) Features estructurales: Clasificación canónica
# -----------------------------
# Esta taxonomía sigue una clasificación bíblica estándar y es útil para:
# - chunking adaptativo más fino
# - análisis de sesgo por grupos canónicos
# - control del retrieval (metadatos)

CANON_GROUP_MAP_ES = {
    # Antiguo Testamento — Pentateuco (Ley)
    "Génesis": "pentateuco",
    "Éxodo": "pentateuco",
    "Levítico": "pentateuco",
    "Números": "pentateuco",
    "Deuteronomio": "pentateuco",

    # Antiguo Testamento — Históricos
    "Josué": "historicos",
    "Jueces": "historicos",
    "Rut": "historicos",
    "1 Samuel": "historicos",
    "2 Samuel": "historicos",
    "1 Reyes": "historicos",
    "2 Reyes": "historicos",
    "1 Crónicas": "historicos",
    "2 Crónicas": "historicos",
    "Esdras": "historicos",
    "Nehemías": "historicos",
    "Ester": "historicos",

    # Antiguo Testamento — Poéticos / Sabiduría
    "Job": "poeticos_sabiduria",
    "Salmos": "poeticos_sabiduria",
    "Proverbios": "poeticos_sabiduria",
    "Eclesiastés": "poeticos_sabiduria",
    "Cantares": "poeticos_sabiduria",
    "Cantar de los Cantares": "poeticos_sabiduria",

    # Antiguo Testamento — Profetas mayores
    "Isaías": "profetas_mayores",
    "Jeremías": "profetas_mayores",
    "Lamentaciones": "profetas_mayores",
    "Ezequiel": "profetas_mayores",
    "Daniel": "profetas_mayores",

    # Antiguo Testamento — Profetas menores
    "Oseas": "profetas_menores",
    "Joel": "profetas_menores",
    "Amós": "profetas_menores",
    "Abdías": "profetas_menores",
    "Jonás": "profetas_menores",
    "Miqueas": "profetas_menores",
    "Nahúm": "profetas_menores",
    "Habacuc": "profetas_menores",
    "Sofonías": "profetas_menores",
    "Hageo": "profetas_menores",
    "Zacarías": "profetas_menores",
    "Malaquías": "profetas_menores",

    # Nuevo Testamento — Evangelios
    "Mateo": "evangelios",
    "Marcos": "evangelios",
    "Lucas": "evangelios",
    "Juan": "evangelios",

    # Nuevo Testamento — Libro histórico (Hechos)
    "Hechos de los Apóstoles": "hechos",

    # Nuevo Testamento — Cartas paulinas (13)
    "Romanos": "cartas_paulinas",
    "1 Corintios": "cartas_paulinas",
    "2 Corintios": "cartas_paulinas",
    "Gálatas": "cartas_paulinas",
    "Efesios": "cartas_paulinas",
    "Filipenses": "cartas_paulinas",
    "Colosenses": "cartas_paulinas",
    "1 Tesalonicenses": "cartas_paulinas",
    "2 Tesalonicenses": "cartas_paulinas",
    "1 Timoteo": "cartas_paulinas",
    "2 Timoteo": "cartas_paulinas",
    "Tito": "cartas_paulinas",
    "Filemón": "cartas_paulinas",

    # Nuevo Testamento — Cartas generales (8) + Hebreos (la tratamos como general para evitar debate de autoría)
    "Hebreos": "cartas_generales",
    "Santiago": "cartas_generales",
    "1 Pedro": "cartas_generales",
    "2 Pedro": "cartas_generales",
    "1 Juan": "cartas_generales",
    "2 Juan": "cartas_generales",
    "3 Juan": "cartas_generales",
    "Judas": "cartas_generales",

    # Nuevo Testamento — Apocalipsis
    "Apocalipsis": "apocalipsis",
}

df["canon_group"] = df["book_es"].map(CANON_GROUP_MAP_ES).fillna("otro")

print("\nDistribución por canon_group:")
print(df["canon_group"].value_counts())

# Control de calidad: libros no mapeados:
unmapped = sorted(df.loc[df["canon_group"] == "otro", "book_es"].unique().tolist())
print("\nLibros no mapeados explícitamente:", unmapped)

# Versión "coarse" para análisis más general
CANON_GROUP_COARSE = {
    "pentateuco": "AT_ley",
    "historicos": "AT_historia",
    "poeticos_sabiduria": "AT_poesia_sabiduria",
    "profetas_mayores": "AT_profetas",
    "profetas_menores": "AT_profetas",
    "evangelios": "NT_evangelios",
    "hechos": "NT_historia",
    "cartas_paulinas": "NT_cartas",
    "cartas_generales": "NT_cartas",
    "apocalipsis": "NT_apocaliptico",
    "otro": "otro"
}

df["canon_group_coarse"] = df["canon_group"].map(CANON_GROUP_COARSE).fillna("otro")

# -----------------------------
# 3) Features numéricas auxiliares (longitudes)
# -----------------------------
WORD_RE = re.compile(r"\b\w+\b", flags=re.UNICODE)

def word_count(text: str) -> int:
    if not text:
        return 0
    return len(WORD_RE.findall(text))

df["wc_rv1909"] = df["text_rv1909"].map(word_count)
df["wc_vbl"] = df["text_vbl"].map(word_count)

# Diferencia de longitudes entre traducciones
df["wc_diff_vbl_minus_rv"] = df["wc_vbl"] - df["wc_rv1909"]

# -----------------------------
# 4) Discretización (binning) de longitudes
# -----------------------------
bins = [-1, 8, 16, 32, 64, 128, 10_000]
labels = ["muy_corto", "corto", "medio", "largo", "muy_largo", "extremo"]

df["len_bin_rv1909"] = pd.cut(df["wc_rv1909"], bins=bins, labels=labels)
df["len_bin_vbl"] = pd.cut(df["wc_vbl"], bins=bins, labels=labels)

df[[
    "ref", "book_es", "testament", "canon_group", "canon_group_coarse",
    "wc_rv1909", "wc_vbl", "wc_diff_vbl_minus_rv",
    "len_bin_rv1909", "len_bin_vbl"
]].sample(5, random_state=42)
```

### **2.3 Chunking adaptativo (pasajes) — Feature Engineering para RAG**

En esta sección se implementa un **chunking adaptativo**, cuyo objetivo es transformar versículos individuales en **unidades semánticas recuperables** (pasajes), adecuadas para un sistema de Recuperación Aumentada por Generación (RAG).

A partir de los hallazgos del EDA y de la clasificación canónica construida en la sección anterior, se define una estrategia de segmentación que **no es global ni uniforme**, sino que se ajusta al tipo de contenido bíblico. Esta decisión responde a que distintos géneros presentan densidades semánticas y estructuras discursivas diferentes.

El chunking adaptativo busca:
- Preservar coherencia semántica.
- Evitar fragmentación excesiva en textos densos.
- Reducir el sesgo de recuperación hacia libros largos.
- Preparar el texto para la generación de embeddings y búsqueda semántica.

```python
# ============================================================
# Parte 2.3 — Chunking adaptativo (pasajes) para RAG
# ============================================================

# -----------------------------
# 1) Reglas de chunking por grupo canónico
# -----------------------------
# Valores expresados como número de versículos por chunk.
# Se elige un valor medio dentro de un rango razonable por grupo.

CHUNK_SIZE_BY_CANON_GROUP = {
    "pentateuco": 4,
    "historicos": 4,
    "poeticos_sabiduria": 2,
    "profetas_mayores": 4,
    "profetas_menores": 3,
    "evangelios": 3,
    "hechos": 4,
    "cartas_paulinas": 3,
    "cartas_generales": 3,
    "apocalipsis": 2,
    "otro": 3,
}

def get_chunk_size(canon_group: str) -> int:
    """Devuelve el tamaño de chunk según el grupo canónico."""
    return CHUNK_SIZE_BY_CANON_GROUP.get(canon_group, 3)

# -----------------------------
# 2) Preparación del dataframe
# -----------------------------
# Orden canónico + orden interno
df_sorted = df.sort_values(
    ["book", "chapter", "verse"]
).reset_index(drop=True)

# -----------------------------
# 3) Construcción de chunks
# -----------------------------
chunks = []
chunk_id = 0

for (book, chapter), g in tqdm(
    df_sorted.groupby(["book", "chapter"], sort=False),
    desc="Construyendo chunks adaptativos"
):
    g = g.sort_values("verse").reset_index(drop=True)

    canon_group = g["canon_group"].mode().iloc[0]
    testament = g["testament"].mode().iloc[0]
    book_es = g["book_es"].mode().iloc[0]
    chunk_size = get_chunk_size(canon_group)

    for start in range(0, len(g), chunk_size):
        block = g.iloc[start:start + chunk_size]

        verse_start = int(block["verse"].iloc[0])
        verse_end = int(block["verse"].iloc[-1])

        if verse_start == verse_end:
            chunk_ref = f"{book} {chapter}:{verse_start}"
        else:
            chunk_ref = f"{book} {chapter}:{verse_start}-{verse_end}"

        chunks.append({
            "chunk_id": f"chk_{chunk_id:06d}",
            "book": book,
            "book_es": book_es,
            "testament": testament,
            "canon_group": canon_group,
            "canon_group_coarse": block["canon_group_coarse"].mode().iloc[0],
            "chapter": int(chapter),
            "verse_start": verse_start,
            "verse_end": verse_end,
            "chunk_ref": chunk_ref,
            "chunk_text_rv1909": " ".join(block["text_rv1909"].tolist()),
            "chunk_text_vbl": " ".join(block["text_vbl"].tolist()),
        })

        chunk_id += 1

df_chunks = pd.DataFrame(chunks)

print("Número total de chunks creados:", df_chunks.shape[0])
df_chunks.sample(5, random_state=42)
```

#### **Interpretación del resultado (2.3)**

En esta sección se implementó una estrategia de chunking adaptativo, transformando el corpus bíblico en unidades semánticas coherentes (pasajes) adecuadas para un sistema de Recuperación Aumentada por Generación (RAG).

La segmentación se basó en la clasificación canónica previamente construida, permitiendo ajustar el tamaño de los chunks según la densidad semántica de cada grupo bíblico. Como resultado, se generaron 10,058 chunks, cada uno con contexto suficiente, trazabilidad canónica y continuidad textual.

Este proceso reduce el sesgo estructural, mejora la recuperación semántica y prepara el corpus para la extracción de características densas mediante embeddings.

### **2.4 Codificación de variables categóricas y extracción de características semánticas (Embeddings)**

En esta sección se realiza la extracción de características semánticas densas (embeddings) a partir de los chunks construidos previamente.

Mientras que las variables estructurales aportan contexto y control, los embeddings permiten representar el significado profundo del texto en un espacio vectorial continuo, facilitando la comparación semántica entre pasajes bíblicos y consultas del usuario.

Asimismo, se codifican variables categóricas relevantes para su posible uso en modelos híbridos o análisis posteriores. Los embeddings se generan de manera separada para cada traducción, preservando diferencias estilísticas y semánticas entre versiones.

```python
# ============================================================
# Parte 2.4 — Codificación categórica y embeddings semánticos
# ============================================================

# -----------------------------
# 1) Codificación de variables categóricas (One-Hot)
# -----------------------------
# Nota: no se aplican a los embeddings directamente, sino como features auxiliares

categorical_cols = ["testament", "canon_group", "canon_group_coarse"]
df_chunks_encoded = pd.get_dummies(
    df_chunks,
    columns=categorical_cols,
    prefix=categorical_cols
)

print("Shape con variables categóricas codificadas:", df_chunks_encoded.shape)

# Guardamos versión tabular (sin embeddings)
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
tabular_path = f"{ARTIFACTS_DIR}/lumina_chunks_tabular_{timestamp}.parquet"
df_chunks_encoded.to_parquet(tabular_path, index=False)

print("Dataset tabular guardado en:", tabular_path)

# -----------------------------
# 2) Extracción de embeddings semánticos
# -----------------------------
# Usamos un modelo multilingüe robusto, adecuado para español y retrieval semántico

MODEL_NAME = "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
model = SentenceTransformer(MODEL_NAME)

def embed_texts(texts, batch_size=64):
    return model.encode(
        texts,
        batch_size=batch_size,
        show_progress_bar=True,
        convert_to_numpy=True,
        normalize_embeddings=True  # útil para similitud coseno
    )

# Textos por traducción
texts_rv = df_chunks["chunk_text_rv1909"].tolist()
texts_vbl = df_chunks["chunk_text_vbl"].tolist()

# Generación de embeddings
emb_rv = embed_texts(texts_rv)
emb_vbl = embed_texts(texts_vbl)

print("Embeddings RV1909:", emb_rv.shape)
print("Embeddings VBL:", emb_vbl.shape)

# -----------------------------
# 3) Guardado de embeddings
# -----------------------------
embeddings_path = f"{EMB_DIR}/lumina_embeddings_{timestamp}.npz"
np.savez_compressed(
    embeddings_path,
    chunk_id=df_chunks["chunk_id"].to_numpy(),
    emb_rv1909=emb_rv,
    emb_vbl=emb_vbl
)

print("Embeddings guardados en:", embeddings_path)

# Manifest de trazabilidad
manifest = {
    "model": MODEL_NAME,
    "normalized": True,
    "n_chunks": int(df_chunks.shape[0]),
    "created_at": timestamp
}

manifest_path = f"{EMB_DIR}/embedding_manifest_{timestamp}.json"
with open(manifest_path, "w", encoding="utf-8") as f:
    json.dump(manifest, f, ensure_ascii=False, indent=2)

print("Manifest guardado en:", manifest_path)
```

#### **Conclusión final de la Parte 2 – Generación de características**

En esta parte se llevó a cabo la generación completa de características para el corpus bíblico de LUMINA, transformando el texto original en representaciones estructuradas y semánticas adecuadas para aprendizaje automático y recuperación de información.

Se construyeron unidades semánticas coherentes (chunks) mediante una estrategia de segmentación adaptativa basada en la clasificación canónica del texto, logrando un equilibrio entre contexto, granularidad y trazabilidad. Asimismo, se codificaron variables categóricas relevantes y se extrajeron embeddings semánticos densos para cada pasaje, utilizando un modelo multilingüe robusto y eficiente.

Como resultado, se generaron artefactos persistentes y reproducibles que representan el conocimiento bíblico tanto en forma tabular como vectorial, dejando el dataset completamente preparado para tareas posteriores de retrieval semántico y modelado dentro de un sistema RAG.

# **Parte 3 – Normalización y transformación**

En esta sección se analizan las decisiones relacionadas con la **normalización, escalamiento y transformación de características**, en el contexto de un proyecto de procesamiento de lenguaje natural basado en embeddings.

A diferencia de datasets puramente numéricos, en LUMINA coexisten distintos tipos de características: variables semánticas densas (embeddings), variables numéricas auxiliares y variables categóricas codificadas. Por ello, no todas las transformaciones comunes son aplicables ni deseables.

El objetivo de esta parte es **justificar de forma explícita**:
- qué características se normalizan o escalan,
- cuáles se dejan sin transformar,
- y por qué estas decisiones favorecen la estabilidad, equidad y convergencia de modelos posteriores.

#### **¿Qué se escala y qué no?**

**No se escalan:**

**- Embeddings semánticos**
  - Ya están normalizados ('normalize_embeddings'=True)

En este avance **no se aplica escalamiento explícito**, porque:

- no se entrenan modelos aún,

- el escalamiento se ejecuta típicamente *justo antes del entrenamiento*.

#### **Transformaciones (log, Box-Cox, etc.)**

No se aplican transformaciones no lineales al texto ni a los embeddings, ya que:

- el texto ya fue proyectado a un espacio latente semántico,
- las transformaciones podrían distorsionar similitudes semánticas,
- y no aportan beneficios claros en esta fase.

Las transformaciones quedan **fuera del alcance del Avance 2** y se considerarían únicamente en contextos de modelado supervisado posterior.

# **Parte 4 – Selección y extracción de características**

En esta sección se abordan las decisiones relacionadas con la selección y extracción de características, con el objetivo de reducir complejidad, mitigar sesgos y asegurar que el conjunto final de variables sea adecuado para tareas posteriores de recuperación semántica y modelado.

En el contexto de procesamiento de lenguaje natural moderno, la selección de características difiere sustancialmente de enfoques clásicos basados en variables tabulares independientes. En *LUMINA*, el núcleo informativo del sistema reside en representaciones semánticas densas (embeddings), mientras que las características estructurales y numéricas cumplen un rol complementario de control, balance y trazabilidad.

Por ello, esta parte se centra tanto en justificar por qué ciertos métodos clásicos no son apropiados, como en explicar por qué los embeddings constituyen una forma legítima y poderosa de extracción de características.

#### **4.1 Selección de características: por qué NO aplicar filtros estadísticos clásicos**

**Explicación conceptual**

En problemas tradicionales de machine learning supervisado, es común aplicar técnicas de selección de características como:

- Umbral de varianza
- Correlación entre variables
- Pruebas Chi-cuadrado
- ANOVA

Sin embargo, estas técnicas **no son adecuadas** para el núcleo semántico de *LUMINA* por las siguientes razones:

1. **Los embeddings no representan variables independientes**   
Cada dimensión de un embedding no tiene un significado semántico aislado; el significado emerge de la *relación conjunta entre dimensiones*.

2. **No existe una variable objetivo en esta fase**   
El proyecto se encuentra aún en la fase de Data Preparation dentro de CRISP-ML(Q). No hay etiquetas ni tareas supervisadas que permitan evaluar relevancia estadística clásica.

3. **Aplicar filtros sobre embeddings rompería la geometría semántica**   
Eliminar dimensiones por varianza o correlación puede distorsionar las distancias angulares, afectando negativamente la similitud coseno utilizada en retrieval.

Por estas razones, **no se aplican técnicas de selección de características clásicas sobre los embeddings**, y esta omisión es una **decisión metodológica consciente**, no una limitación técnica.

**¿Qué sí se selecciona explícitamente?**

La selección de características en *LUMINA* se realiza **a nivel de tipos de variables**, no de dimensiones individuales:

- **Se conservan:**
  - Embeddings semánticos completos (RV1909 y VBL).
  - Metadatos canónicos ('testament', 'canon_group', etc.).

- **Se consideran auxiliares:**
  - Longitudes y bins, útiles solo en modelos híbridos o análisis posteriores.

- **Se excluyen del modelado semántico directo:**
  - Variables puramente descriptivas que no aportan significado.

#### **4.2 Extracción de características: embeddings como representación latente**

**Explicación conceptual**

La extracción de características consiste en transformar los datos originales en una representación más compacta y significativa. En *LUMINA*, esta transformación se realizó mediante **embeddings semánticos**, que proyectan cada chunk bíblico a un espacio vectorial continuo de baja dimensionalidad.

Desde una perspectiva metodológica, los embeddings cumplen el mismo rol que técnicas clásicas de extracción como PCA o análisis factorial, pero con ventajas clave:

- Capturan **significado contextual**, no solo varianza.
- Preservan relaciones semánticas no lineales.
- Son independientes del idioma superficial del texto.

Por tanto, el uso de embeddings constituye una **técnica de extracción de características avanzada**, alineada con el estado del arte en NLP.

#### **4.3 PCA exploratorio sobre embeddings (análisis, no reducción)**

**Justificación**

Aunque **no se utiliza PCA para reducir dimensionalidad en producción**, puede aplicarse de forma exploratoria para:

- Verificar que los embeddings no colapsan.
- Observar separación semántica general entre grupos canónicos.
- Detectar anomalías evidentes.

Este análisis **no modifica el dataset**, solo sirve como validación cualitativa.

```python
# ============================================================
# Parte 4.3 — PCA exploratorio sobre embeddings (análisis)
# ============================================================

from sklearn.decomposition import PCA

# Usamos una submuestra para evitar costos computacionales altos
sample_size = 2000
rng = np.random.default_rng(42)

idx = rng.choice(len(emb_rv), size=sample_size, replace=False)
emb_sample = emb_rv[idx]

pca = PCA(n_components=2, random_state=42)
emb_2d = pca.fit_transform(emb_sample)

print("Varianza explicada por los 2 primeros componentes:")
print(pca.explained_variance_ratio_)
print("Varianza acumulada:", pca.explained_variance_ratio_.sum())
```

#### **Interpretación**

La varianza explicada por los dos primeros componentes principales (~17.9 %) es consistente con lo esperado en representaciones semánticas densas. En embeddings de lenguaje natural, el significado se distribuye a lo largo de múltiples dimensiones, por lo que una varianza acumulada moderada en las primeras componentes indica diversidad semántica y ausencia de colapso representacional.

Estos resultados sugieren que los embeddings capturan información rica y no lineal, confirmando que el espacio vectorial es adecuado para tareas de similitud semántica y recuperación de información, sin justificar una reducción de dimensionalidad agresiva.

#### **4.4 Decisiones conscientes de selección y extracción**

A partir del análisis exploratorio realizado, se tomaron las siguientes decisiones metodológicas conscientes:

- **No aplicar reducción de dimensionalidad sobre los embeddings**  
El análisis PCA mostró que las primeras componentes explican una fracción moderada de la varianza total, lo cual es característico de representaciones semánticas distribuidas. Reducir dimensionalidad podría eliminar información semántica relevante para la recuperación.

- **Preservar el espacio semántico completo para retrieval**  
Mantener los embeddings completos permite conservar relaciones semánticas finas entre pasajes bíblicos, fundamentales para sistemas RAG basados en similitud coseno.

- **Usar PCA únicamente con fines diagnósticos**  
El PCA se utilizó como herramienta exploratoria para verificar la estabilidad y diversidad del espacio vectorial, no como mecanismo de compresión o selección de características.

- **Separar claramente análisis exploratorio de decisiones productivas**  
Esta distinción garantiza reproducibilidad, trazabilidad y alineación con la fase de Data Preparation definida por CRISP-ML(Q).

Estas decisiones priorizan la fidelidad semántica del corpus y sientan una base sólida para las fases posteriores de modelado y evaluación.
