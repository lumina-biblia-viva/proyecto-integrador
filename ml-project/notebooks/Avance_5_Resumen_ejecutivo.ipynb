### **3. Resultados Comparativos**

```python
# ==========================================
# Carga CSV, ordena resultados e imprime la tabla
# ==========================================

import pandas as pd

# Cargar resultados y copiar tabla
comparison = pd.read_csv("comparison_modelos_avance5.csv")
executive_table = comparison.copy()

# Redondear métricas para presentación
cols_round = [
    "judge_overall_adj_mean",
    "leakage_rate_mean",
    "total_time_mean"
]

for c in cols_round:
    if c in executive_table.columns:
        executive_table[c] = executive_table[c].astype(float).round(3)

# Ordenar por calidad
executive_table = executive_table.sort_values(
    by="judge_overall_adj_mean",
    ascending=False
).reset_index(drop=True)

# Función para resaltar la primera fila (ensamble ganador)
def highlight_first_row(row):
    if row.name == 0:
        return ['background-color: lightgreen'] * len(row)
    else:
        return [''] * len(row)

# Mostrar tabla estilizada
executive_table.style.apply(highlight_first_row, axis=1)
```
